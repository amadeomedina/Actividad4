### Errores comunes a evitar

| Error | Consecuencia | Solución |
|-------|-------------|----------|
| API key en el código | Seguridad comprometida, repo bloqueado | Usar secrets de HF/Streamlit + `os.environ` |
| No testear antes del deploy | "En mi notebook funcionaba" | Probar en un entorno limpio |
| README vacío | El evaluador no sabe qué hace la app | Escribir el README antes del código |
| Ignorar errores de la API | La app se rompe cuando la API falla | `try/except` en todas las llamadas a APIs |
| Proyecto demasiado ambicioso | No se termina a tiempo | Empezar con lo mínimo funcional |

````python
import os
import gradio as gr
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools import DuckDuckGoSearchRun

# 1. Configuración Segura del Modelo (Cero API Keys en código)
try:
    # Se recupera el secreto desde el gestor de Colab
    api_key = userdata.get('GOOGLE_API_KEY')
    if not api_key:
        llm = None
        config_error = "⚠️ Error: No se encontró 'GOOGLE_API_KEY' en los Secretos de Colab."
    else:
        os.environ["GOOGLE_API_KEY"] = api_key
        llm = ChatGoogleGenerativeAI(model="models/gemini-flash-latest", temperature=0.7)
        config_error = None
except Exception as e:
    llm = None
    config_error = f"⚠️ Error crítico de configuración: {str(e)}"

# 2. Herramientas con Aislamiento de Errores
try:
    from duckduckgo_search import DDGS
    import sys
    class Dummy: pass
    ddgs_module = Dummy()
    ddgs_module.DDGS = DDGS
    sys.modules["ddgs"] = ddgs_module
    search = DuckDuckGoSearchRun()
except Exception:
    search = None

def calculator(expression):
    try:
        allowed = "0123456789+-*/(). "
        clean_expr = "".join([c for c in expression if c in allowed])
        if not clean_expr.strip(): return "Expresión no válida."
        # Uso de eval limitado para mayor seguridad en este contexto
        return str(eval(clean_expr, {"__builtins__": None}, {}))
    except ZeroDivisionError:
        return "Error: División por cero."
    except Exception as e:
        return f"Error matemático: {str(e)}"

# 3. Gestión de Memoria Multisesión Estricta
class ChatSession:
    def __init__(self, id):
        self.id = id
        self.name = f"Conversación {id}"
        self.history = []
        self.gradio_format = []
        self.response_count = 0

    def add(self, role, content):
        self.history.append(f"{role}: {content}")
        if role == "Asistente":
            self.response_count += 1

# Estado Global
sessions = {1: ChatSession(1)}

def predict(message, session_id):
    # Validaciones de entrada y configuración
    if config_error:
        gr.Error(config_error)
        return "", sessions[session_id].gradio_format, sessions[session_id].response_count

    if not message or not message.strip():
        gr.Warning("Por favor, escribe un mensaje.")
        return "", sessions[session_id].gradio_format, sessions[session_id].response_count

    session = sessions[session_id]
    lower_msg = message.lower()
    context_info = ""

    # Ejecución segura de herramientas
    try:
        if any(word in lower_msg for word in ["busca", "quien", "que es", "noticias"]):
            context_info = f"\n[Búsqueda]: {search.run(message)}" if search else "\n[Aviso]: Buscador no disponible."
        elif any(char in lower_msg for char in ["+", "*", "/", "-"]) and any(i.isdigit() for i in lower_msg):
            context_info = f"\n[Calculadora]: {calculator(message)}"
    except Exception as e:
        context_info = f"\n[Herramienta omitida por error]: {str(e)}"

    full_prompt = f"Eres ChatMedina, un asistente seguro y avanzado.\nHistorial:\n{'\n'.join(session.history[-6:])}\n\nContexto: {context_info}\n\nUsuario: {message}\nRespuesta:"

    try:
        response = llm.invoke(full_prompt)
        answer = response.content if isinstance(response.content, str) else "".join([it['text'] for it in response.content if 'text' in it])
    except Exception as e:
        answer = "Hubo un problema procesando tu solicitud. Por favor, reintenta."
        gr.Error(f"Error de API: {str(e)}")

    session.add("Usuario", message)
    session.add("Asistente", answer)
    session.gradio_format.append({"role": "user", "content": message})
    session.gradio_format.append({"role": "assistant", "content": answer})

    return "", session.gradio_format, session.response_count

# 4. Interfaz Protegida
custom_theme = gr.themes.Soft(primary_hue="blue", secondary_hue="sky").set(
    body_background_fill="#e3f2fd",
    block_background_fill="#ffffff"
)

with gr.Blocks(theme=custom_theme, title="🚀 ChatMedina Pro") as demo:
    active_session_state = gr.State(1)
    gr.Markdown("# 🚀 ChatMedina Pro (Versión Segura)")
    if config_error: gr.Markdown(f"### {config_error}")

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("### 📁 Mis Chats")
            new_chat_btn = gr.Button("+ Nuevo Chat", variant="primary")
            session_list = gr.Radio(
                choices=[s.name for s in sessions.values()],
                value=sessions[1].name,
                label="Historial"
            )
            with gr.Group():
                gr.Markdown("### 📊 Estadísticas")
                counter_display = gr.Number(label="Respuestas", value=0, interactive=False)

        with gr.Column(scale=3):
            chatbot_ui = gr.Chatbot(label="Conversación", height=550)
            msg_input = gr.Textbox(label="Entrada", placeholder="Escribe aquí...")
            send_btn = gr.Button("Enviar", variant="primary")

    def load_session(name):
        try:
            sid = next(s.id for s in sessions.values() if s.name == name)
            return sid, sessions[sid].gradio_format, sessions[sid].response_count
        except: return 1, [], 0

    def create_new():
        new_id = max(sessions.keys()) + 1
        sessions[new_id] = ChatSession(new_id)
        return gr.update(choices=[s.name for s in sessions.values()], value=sessions[new_id].name), new_id, [], 0

    msg_input.submit(predict, [msg_input, active_session_state], [msg_input, chatbot_ui, counter_display])
    send_btn.click(predict, [msg_input, active_session_state], [msg_input, chatbot_ui, counter_display])
    session_list.change(load_session, session_list, [active_session_state, chatbot_ui, counter_display])
    new_chat_btn.click(create_new, None, [session_list, active_session_state, chatbot_ui, counter_display])

if __name__ == "__main__":
    demo.launch(debug=False)
````
